```python
import time
import re


# ---------------- Extended Semantic Emoji Dictionary ---------------- #
WORD_TO_EMOJI = {
    # ❤️ Emotions & Relationships
    "love": "❤️", "like": "👍", "happy": "😊", "joy": "😁", "sad": "😢",
    "angry": "😡", "fear": "😨", "surprise": "😲", "excited": "🤩",
    "kiss": "😘", "hug": "🤗", "heart": "❤️", "honey": "🍯", "sweet": "🍬",

    # 👨‍💻 Tech
    "ai": "🤖", "robot": "🤖", "computer": "💻", "laptop": "💻",
    "network": "📡", "internet": "🌐", "data": "📊", "system": "⚙️",
    "security": "🔐", "hack": "💀", "code": "👨‍💻", "program": "🧠",

    # 💰 Finance
    "money": "💰", "bank": "🏦", "cash": "💵", "payment": "💳",
    "crypto": "🪙", "bitcoin": "₿", "stock": "📈", "loss": "📉",
    "profit": "💹", "salary": "💼",

    # 🌍 Daily Life
    "home": "🏠", "house": "🏠", "food": "🍔", "pizza": "🍕",
    "coffee": "☕", "tea": "🍵", "water": "💧", "car": "🚗",
    "travel": "✈️", "flight": "✈️", "phone": "📱",

    # 🔥 Nature & Objects
    "fire": "🔥", "star": "🌟", "sun": "☀️", "moon": "🌙",
    "tree": "🌳", "flower": "🌸", "rain": "🌧️",

    # 🧍 Actions
    "run": "🏃", "walk": "🚶", "eat": "🍽️", "sleep": "😴",
    "work": "💼", "study": "📚", "read": "📖", "write": "✍️",

    # 👥 People
    "man": "👨", "woman": "👩", "boy": "👦", "girl": "👧",
    "friend": "🤝", "family": "👨‍👩‍👧‍👦"
}

# Reverse mapping (1:1)
EMOJI_TO_WORD = {v: k for k, v in WORD_TO_EMOJI.items()}


# ---------------- Phrase Mapping (Optional Enhancement) ---------------- #
PHRASE_TO_EMOJI = {
    "i love you": "❤️❤️",
    "good morning": "🌅",
    "good night": "🌙",
    "thank you": "🙏",
    "well done": "👏"
}

EMOJI_TO_PHRASE = {v: k for k, v in PHRASE_TO_EMOJI.items()}


# ---------------- Tokenization ---------------- #
def tokenize(text):
    # Split words and punctuation separately
    return re.findall(r"\w+|[^\w\s]", text)


# ---------------- Encode ---------------- #
def encode(text):
    text_lower = text.lower()

    # Phrase-level replacement first
    for phrase, emoji in PHRASE_TO_EMOJI.items():
        text_lower = text_lower.replace(phrase, emoji)

    tokens = tokenize(text_lower)
    result = []

    for token in tokens:
        if token in WORD_TO_EMOJI:
            result.append(WORD_TO_EMOJI[token])
        else:
            result.append(token)

    return " ".join(result)


# ---------------- Decode ---------------- #
def decode(cipher):
    tokens = cipher.split()
    result = []

    for token in tokens:
        if token in EMOJI_TO_WORD:
            result.append(EMOJI_TO_WORD[token])
        elif token in EMOJI_TO_PHRASE:
            result.append(EMOJI_TO_PHRASE[token])
        else:
            result.append(token)

    return " ".join(result)


# ---------------- Main ---------------- #
def main():
    print("=== Advanced Semantic Emoji Encryption System ===")

    text = input("Enter text: ").strip()

    # Encode
    cipher = encode(text)

    print("\n🎭 Emoji Output:")
    print(cipher)

    duration = int(input("\nEnter time duration (seconds): "))

    print(f"\n⏳ Waiting {duration} seconds...")
    time.sleep(duration)

    # Decode
    original = decode(cipher)

    print("\n🔓 Time expired! Revealing Original Text:")
    print(original)


if __name__ == "__main__":
    main()
```
